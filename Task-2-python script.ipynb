```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
# Import and Quality check of data
df = pd.read_csv('/content/cleaned_dataset.csv')
df.head()
df.shape
df.describe()
df.isnull().sum()
df.nunique()
```

```python
#Summary statistics of data
print(df.describe())
print(df.describe(exclude='number').T)
print(df['TotalPrice'].skew())
```

```python
#Measure of central tendacy

num =["Quantity","UnitPrice","ItemsInCart","TotalPrice"]
print(df[num].agg(["mean","median"]).round(2))
print(df[num].mode().iloc[0])
print(df[["Product","OrderStatus"]].mode)
```

```python
# Categorical count
obj_count = ["Product","PaymentMethod","OrderStatus","CouponCode","ReferralSource"]
#count percentage for categorical column
counts=df[obj_count].value_counts()
summary = pd.DataFrame({
    "counts":counts,
    "percent":(counts/len(df)*100).round(1),
})
print(f"\n==={obj_count}({df[obj_count].nunique()}categories)===")
print(summary)
```

```python
# Correletion and heatmap
num_colms=["Quantity","UnitPrice","ItemsInCart","TotalPrice"]
corr=df[num_colms].corr()
print(corr.round(2))
#heatmap
sns.set_theme(style="white")
plt.figure(figsize=(6,5))
sns.heatmap(corr,annot=True,cmap="coolwarm",vmin=-1,vmax=1,fmt=".2f",square=True)
plt.title("Correlation_heatmap")
plt.tight_layout()
plt.savefig("correlation_heatmap.png",dpi=110)
plt.show()
```

```python
#EDA Graphs
df[num_colms].hist(bins=25,figsize=(11,7))
plt.tight_layout()
plt.savefig("Distribution.png",dpi=110)
plt.show()

#Revenue and cancel/return rate
df["bad"]=df["OrderStatus"].isin(["Cancelled","Retutned"])
fig, ax = plt.subplots(1, 3, figsize=(17, 5))
df.groupby("Product")["TotalPrice"].sum().sort_values().plot.barh(ax=ax[0], title="Revenue by product")
df.groupby("OrderStatus")["TotalPrice"].sum().plot.bar(ax=ax[1], title="Revenue by status", rot=30)
(df.groupby("PaymentMethod")["bad"].mean() * 100).sort_values().plot.barh(ax=ax[2], title="Cancel/return % by payment")
plt.tight_layout()
plt.savefig("revenue and rates.png", dpi=110)
plt.show()
```
